```python
fmt = "%a, %b %d, %Y | %I:%M %p"
ds = datetime(1980, 12, 1, 15, 33)
ds.strftime(fmt)
```

```python
datetime.strptime("2025-06-4", fmt)
```

```python
fmt = "%m-%d-%y %H:%M"
ds.strftime(fmt)
```

```python
fmts = [
    # Mon, Dec 01, 1980 | 03:33 PM
    "%a, %b %d, %Y | %I:%M %p",
    # 12-01-80 15:33
    "%m-%d-%y %H:%M",
]
```

```python
ds_str = "12-01-80 15:33"
ds: datetime | None = None
for fmt in fmts:
    try:
        ds = datetime.strptime(ds_str, fmt)
        ds_str = ds.isoformat()
    except ValueError:
        pass
```

```python
def is_power_of_two(x: int) -> bool:
    if x <= 0:
        return False
    return (x & (x - 1)) == 0
```
